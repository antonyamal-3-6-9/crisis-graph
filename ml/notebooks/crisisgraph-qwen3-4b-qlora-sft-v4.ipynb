# CrisisGraph Qwen3-4B QLoRA — SFT V4

Train a fresh adapter from the same pinned base as V1–V3, on 608 training and 102 validation records with **V3 targets for every field** (`victim_profile` last) and **prompt v3.2** as the system message (spec option a: train and serve with the same prompt). All labels are owner-reviewed; new message texts were written by zero-context agents that never saw the evaluation sets (`ml/data/triage-sft-v4-spec.md`, `triage-sft-v4-export-manifest.json`).

**Runs on Kaggle or Colab.** The full run takes about 3–5 hours on a T4.

- **Kaggle (recommended, runs with the browser closed):** create a private dataset with the two export files, import this notebook, attach the dataset, set Accelerator = GPU T4 and turn Internet on. Do a quick interactive smoke run (`RUN_MODE = "smoke"`), then set `RUN_MODE = "full"` and use *Save Version → Save & Run All (Commit)*. Download the adapter zip from the version's Output tab.
- **Colab:** run with `RUN_MODE = "smoke"`, restart, set `RUN_MODE = "full"` and run again. Checkpoints go to Google Drive, so a disconnect can be resumed by rerunning all cells.

Checkpoints are saved every 19 steps (a quarter epoch); rerunning a stopped full run resumes from the latest one. Never continue training an earlier adapter or the smoke adapter.

**One change from SFT-V3:** `MAX_LENGTH` is 3,072 tokens instead of 1,024, because prompt v3.2 is about 2,400 tokens. Section 3 checks that every record fits; nothing is truncated. If a T4 runs out of memory, stop and record it (spec: the fallback is decided then, not by silently shortening the prompt).

The model extracts candidate facts only. Deterministic Rust services retain routing, verification, allocation, and dispatch authority.

Hash verification prevents accidentally training on an older candidate revision or on the frozen evaluation set.

```python
if not ON_KAGGLE and (not TRAIN_FILE.exists() or not VALIDATION_FILE.exists()):
    from google.colab import files

    print("Upload the exported train and validation JSONL files.")
    uploaded = files.upload()
    for path in [TRAIN_FILE, VALIDATION_FILE]:
        if path.name in uploaded:
            path.write_bytes(uploaded[path.name])

def sha256(path: Path) -> str:
    digest = hashlib.sha256()
    with path.open("rb") as handle:
        for chunk in iter(lambda: handle.read(1024 * 1024), b""):
            digest.update(chunk)
    return digest.hexdigest()

for path in [TRAIN_FILE, VALIDATION_FILE]:
    assert path.exists(), f"Missing required input (upload, or attach the Kaggle dataset): {path.name}"
    actual_hash = sha256(path)
    assert actual_hash == EXPECTED_SHA256[path.name], (
        f"Checksum mismatch for {path.name}: {actual_hash}"
    )
    print(path.name, actual_hash, "verified")
```

## 4. Load the 4-bit base model and inject LoRA adapters

The base weights stay frozen in NF4. Gradients update only the low-rank adapter matrices. This pilot uses FP16 compute across Colab GPU types and retains trainable adapter weights in FP32, avoiding BF16/AMP gradient-scaler incompatibilities across changing Colab package builds.

```python
from peft import LoraConfig, prepare_model_for_kbit_training
from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig

use_bf16 = False
compute_dtype = torch.float16

quantization_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_use_double_quant=True,
    bnb_4bit_compute_dtype=compute_dtype,
)

tokenizer = AutoTokenizer.from_pretrained(MODEL_ID, revision=MODEL_REVISION, use_fast=True)
if tokenizer.pad_token_id is None:
    tokenizer.pad_token = tokenizer.eos_token
tokenizer.padding_side = "right"

model = AutoModelForCausalLM.from_pretrained(
    MODEL_ID,
    revision=MODEL_REVISION,
    quantization_config=quantization_config,
    device_map={"": 0},
    torch_dtype=compute_dtype,
)
model.config.use_cache = False
model = prepare_model_for_kbit_training(
    model,
    use_gradient_checkpointing=True,
)

lora_config = LoraConfig(
    r=16,
    lora_alpha=32,
    lora_dropout=0.05,
    bias="none",
    task_type="CAUSAL_LM",
    target_modules="all-linear",
)
print("Compute dtype:", compute_dtype)

# Every record must fit MAX_LENGTH: a truncated record would cut the target JSON.
lengths = {
    split: [
        len(tokenizer(tokenizer.apply_chat_template(row["messages"], tokenize=False),
                      add_special_tokens=False)["input_ids"])
        for row in dataset[split]
    ]
    for split in ["train", "validation"]
}
token_stats = {split: {"max": max(v), "mean": round(sum(v) / len(v))} for split, v in lengths.items()}
print("Tokenised record lengths:", token_stats)
assert max(max(v) for v in lengths.values()) <= MAX_LENGTH, (
    "A record exceeds MAX_LENGTH; stop and record it instead of truncating."
)
```

## 5. Configure supervised fine-tuning

Loss is computed only on assistant responses, so the model learns the target JSON instead of learning to reproduce the system prompt or SOS message. Packing is disabled for the first controlled run to keep example boundaries simple to inspect.

```python
from trl import SFTConfig, SFTTrainer

training_args = SFTConfig(
    output_dir=str(CHECKPOINT_DIR),
    max_length=MAX_LENGTH,
    packing=False,
    assistant_only_loss=True,
    per_device_train_batch_size=1,
    per_device_eval_batch_size=1,
    gradient_accumulation_steps=8,
    gradient_checkpointing=True,
    learning_rate=1e-4,
    num_train_epochs=2 if RUN_MODE == "full" else 1,
    max_steps=-1 if RUN_MODE == "full" else 3,
    warmup_steps=1 if RUN_MODE == "smoke" else 6,
    lr_scheduler_type="cosine",
    optim="paged_adamw_8bit",
    logging_steps=1 if RUN_MODE == "smoke" else 5,
    # Validation once per epoch, as in SFT-V3. Checkpoints every 19 steps (a quarter
    # epoch) for crash recovery; 19 divides the 76-step epoch, so both epoch-end
    # checkpoints exist and the better one is selected by eval loss after training
    # (load_best_model_at_end would force saving on the evaluation schedule).
    eval_strategy="epoch" if RUN_MODE == "full" else "no",
    save_strategy="steps" if RUN_MODE == "full" else "no",
    save_steps=19,
    save_total_limit=None,
    fp16=not use_bf16,
    bf16=use_bf16,
    seed=SEED,
    data_seed=SEED,
    report_to="none",
)

trainer = SFTTrainer(
    model=model,
    args=training_args,
    train_dataset=dataset["train"],
    eval_dataset=dataset["validation"] if RUN_MODE == "full" else None,
    processing_class=tokenizer,
    peft_config=lora_config,
)

# Some Colab combinations initialize newly injected LoRA weights in the
# base compute dtype. AMP gradient scaling expects trainable master weights
# in FP32, so cast adapters only; the frozen base remains 4-bit NF4.
for parameter in trainer.model.parameters():
    if parameter.requires_grad and parameter.dtype != torch.float32:
        parameter.data = parameter.data.float()

trainable_dtypes = {parameter.dtype for parameter in trainer.model.parameters() if parameter.requires_grad}
assert trainable_dtypes == {torch.float32}, trainable_dtypes
print("Trainable parameter dtypes:", trainable_dtypes)
trainer.model.print_trainable_parameters()
```

```python
batch = next(iter(trainer.get_train_dataloader()))
supervised_tokens = int((batch["labels"] != -100).sum())
all_tokens = int(batch["labels"].numel())
assert supervised_tokens > 0
assert supervised_tokens < all_tokens
print({"supervised_assistant_tokens": supervised_tokens, "batch_tokens": all_tokens})
```

## 6. Train and validate

Do not treat smoke-run loss as an experiment result. For the full run, validation loss selects the best epoch checkpoint; task accuracy is measured later with the frozen 120-case CrisisGraph evaluator.

```python
train_result = trainer.train(resume_from_checkpoint=RESUME_FROM)
train_metrics = dict(train_result.metrics)

best_checkpoint = None
if RUN_MODE == "full":
    # Pick the epoch-end checkpoint with the lower validation loss (SFT-V3 rule).
    epoch_evals = [log for log in trainer.state.log_history if "eval_loss" in log]
    steps_per_epoch = trainer.state.max_steps // int(training_args.num_train_epochs)
    best = min(epoch_evals, key=lambda log: log["eval_loss"])
    best_step = round(best["epoch"] * steps_per_epoch)
    best_checkpoint = CHECKPOINT_DIR / f"checkpoint-{best_step}"
    assert best_checkpoint.exists(), best_checkpoint
    if best_step != trainer.state.global_step:
        from peft import set_peft_model_state_dict
        from safetensors.torch import load_file

        set_peft_model_state_dict(
            trainer.model, load_file(str(best_checkpoint / "adapter_model.safetensors"))
        )
    print("Epoch validation losses:", [(log["epoch"], log["eval_loss"]) for log in epoch_evals])
    print("Selected checkpoint:", best_checkpoint)

eval_metrics = trainer.evaluate() if RUN_MODE == "full" else {}
print("Train metrics:", json.dumps(train_metrics, indent=2, default=str))
print("Validation metrics (selected weights):", json.dumps(eval_metrics, indent=2, default=str))
```

```python
ADAPTER_DIR = OUTPUT_DIR / "adapter"
trainer.save_model(str(ADAPTER_DIR))
tokenizer.save_pretrained(str(ADAPTER_DIR))

run_manifest = {
    "experiment_id": EXPERIMENT_ID,
    "run_mode": RUN_MODE,
    "requested_base_revision": MODEL_REVISION,
    "gpu": torch.cuda.get_device_name(0),
    "cuda_version": torch.version.cuda,
    "base_model": MODEL_ID,
    "base_model_revision": getattr(trainer.model.config, "_commit_hash", None),
    "method": "QLoRA",
    "quantization": {
        "bits": 4,
        "type": "nf4",
        "double_quantization": True,
        "compute_dtype": str(compute_dtype),
    },
    "lora": {
        "rank": 16,
        "alpha": 32,
        "dropout": 0.05,
        "target_modules": "all-linear",
    },
    "training": {
        "seed": SEED,
        "max_length": MAX_LENGTH,
        "tokenised_lengths": token_stats,
        "epochs_requested": 2 if RUN_MODE == "full" else 1,
        "max_steps": -1 if RUN_MODE == "full" else 3,
        "micro_batch_size": 1,
        "gradient_accumulation_steps": 8,
        "learning_rate": 1e-4,
        "warmup_steps": 1 if RUN_MODE == "smoke" else 6,
        "assistant_only_loss": True,
    },
    "data": {
        "dataset_version": "triage-sft-v4",
        "review_policy": REVIEW_POLICY,
        "independent_review_complete": True,
        "prompt_version": "triage-extraction-v3.2",
        "target_schema": "triage-extraction-v3",
        "dev_included": False,
        "train_records": len(dataset["train"]),
        "validation_records": len(dataset["validation"]),
        "sha256": EXPECTED_SHA256,
        "frozen_evaluation_included": False,
    },
    "metrics": {"train": train_metrics, "validation": eval_metrics},
    "best_checkpoint": str(best_checkpoint) if best_checkpoint else None,
    "resumed_from": RESUME_FROM,
    "platform": "kaggle" if ON_KAGGLE else "colab",
    "packages": {
        package: metadata.version(package)
        for package in ["torch", "transformers", "trl", "peft", "accelerate", "bitsandbytes", "datasets"]
    },
}
(OUTPUT_DIR / "run-manifest.json").write_text(
    json.dumps(run_manifest, indent=2, default=str), encoding="utf-8"
)
print("Saved adapter to", ADAPTER_DIR)
print(json.dumps(run_manifest, indent=2, default=str))
```

## 8. Post-training generation smoke check

This checks that the trained adapter can generate a response. It is not the frozen evaluation and must not be reported as accuracy.

```python
system_prompt = dataset["train"][0]["messages"][0]["content"]
sample_messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": "Njangal 3 per Market Hall il, ammakku shwasam muttal und, vellam aravolam. Boat venam."},
]
trained_model = trainer.model
trained_model.gradient_checkpointing_disable()
trained_model.config.use_cache = True
trained_model.eval()
inputs = tokenizer.apply_chat_template(
    sample_messages,
    add_generation_prompt=True,
    tokenize=True,
    return_dict=True,
    return_tensors="pt",
).to(trained_model.device)
input_length = inputs["input_ids"].shape[-1]
with torch.inference_mode():
    generated = trained_model.generate(
        **inputs,
        max_new_tokens=768,
        do_sample=False,
        pad_token_id=tokenizer.pad_token_id,
        eos_token_id=tokenizer.eos_token_id,
    )
response = tokenizer.decode(generated[0, input_length:], skip_special_tokens=True)
print("Adapter response:")
print(response)

with trained_model.disable_adapter(), torch.inference_mode():
    base_generated = trained_model.generate(
        **inputs,
        max_new_tokens=768,
        do_sample=False,
        pad_token_id=tokenizer.pad_token_id,
        eos_token_id=tokenizer.eos_token_id,
    )
base_response = tokenizer.decode(
    base_generated[0, input_length:], skip_special_tokens=True
)
print("\nBase response with adapter disabled:")
print(base_response)
```

## Experimental discipline

A successful training run only proves that optimization completed. The adapter is accepted only if the existing evaluator shows a meaningful improvement in human-review recall and unsupported-fact rate without unacceptable regressions in schema validity, clear-case accuracy, or multilingual behavior. The deterministic routing and verification boundary remains unchanged regardless of the model result.
